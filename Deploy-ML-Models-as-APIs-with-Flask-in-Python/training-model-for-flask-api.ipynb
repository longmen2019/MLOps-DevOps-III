```python
# Import the NumPy library and give it the alias 'np'
# NumPy is used for numerical computations, arrays, and mathematical operations
import numpy as np  

# Import the pandas library and give it the alias 'pd'
# Pandas is used for data manipulation and analysis, especially with DataFrames
import pandas as pd  

# Import the train_test_split function from scikit-learn's model_selection module
# This function is used to split datasets into training and testing sets
from sklearn.model_selection import train_test_split  

# Import the StandardScaler class from scikit-learn's preprocessing module
# StandardScaler standardizes features by removing the mean and scaling to unit variance
from sklearn.preprocessing import StandardScaler  

# Import the LogisticRegression class from scikit-learn's linear_model module
# LogisticRegression is a classification algorithm used for binary or multiclass problems
from sklearn.linear_model import LogisticRegression  

# Import the accuracy_score function from scikit-learn's metrics module
# accuracy_score calculates the proportion of correctly predicted labels in classification tasks
from sklearn.metrics import accuracy_score  

# Import the pickle module from Python's standard library
# Pickle is used to serialize (save) and deserialize (load) Python objects, such as trained models
import pickle  
```

```python
data = pd.read_csv("diabetes.csv")
```

```python
data.head()
```

```python
# Create the feature matrix X by dropping the "Outcome" column from the dataset
# This ensures X contains only the input variables used for training
X = data.drop(columns=["Outcome"])

# Create the target vector y by selecting the "Outcome" column from the dataset
# This is the label we want to predict (e.g., 0 or 1 in a classification task)
y = data["Outcome"]
```

```python
X.head()
```

```python
y
```

```python
# Split the dataset into training and testing subsets
# X = feature matrix, y = target labels
# test_size=0.2 → 20% of the data will be used for testing, 80% for training
# stratify=y → ensures the class distribution (e.g., 0s and 1s) is preserved in both train and test sets
# random_state=42 → sets a fixed random seed for reproducibility (same split every run)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
# Print the shape (number of rows and columns) of the target vector y
# y.shape → shows the total number of samples in the dataset

# Print the shape of y_train → number of samples used for training
# Print the shape of y_test → number of samples used for testing
print(y.shape, y_train.shape, y_test.shape)
```

```python
# Initialize the StandardScaler
# StandardScaler standardizes features by removing the mean and scaling to unit variance
scaler = StandardScaler()

# Fit the scaler on the training data and transform it
# .fit_transform() learns the mean and standard deviation from X_train,
# then applies the scaling so each feature has mean=0 and variance=1
X_train_scaled = scaler.fit_transform(X_train)

# Transform the test data using the same scaling parameters learned from X_train
# .transform() applies the scaling without refitting, ensuring no data leakage
X_test_scaled = scaler.transform(X_test)
```

```python
# Initialize the LogisticRegression model
# Logistic Regression is a classification algorithm commonly used for binary outcomes (e.g., 0 vs 1)
model = LogisticRegression()

# Train (fit) the model using the scaled training features and their corresponding labels
# X_train_scaled → standardized feature matrix
# y_train → target labels
model.fit(X_train_scaled, y_train)
```

```python
# Use the trained model to make predictions on the scaled test set
y_pred = model.predict(X_test_scaled)

# Calculate the accuracy score by comparing predicted labels (y_pred) with true labels (y_test)
accuracy = accuracy_score(y_test, y_pred)

# Print the accuracy rounded to 2 decimal places
print("accuracy", round(accuracy, 2))
```

```python
y_pred_train = model.predict(X_train_scaled)
accuracy = accuracy_score(y_train, y_pred_train)
print("Training accuracy:", round(accuracy, 2))
```

```python
# Save the trained model
with open("diabetes_model.pkl", "wb") as model_file:
    pickle.dump(model, model_file)

# Save the fitted scaler
with open("scaler.pkl", "wb") as scaler_file:
    pickle.dump(scaler, scaler_file)
```

### **Code for Prediction**

```python
import pickle
import numpy as np 

# Load the trained model
with open("diabetes_model.pkl", "rb") as model_file:   
    loaded_model = pickle.load(model_file)

# Load the scaler
with open("scaler.pkl", "rb") as scaler_file:          
    loaded_scaler = pickle.load(scaler_file)
```

```python
new_data_point = [6,148,72,35,0,33.6,0.627,50]

# Wrap in an extra list so it's 2D (scikit-learn expects [n_samples, n_features])
scaled_data = loaded_scaler.transform([new_data_point])

# Now you can feed it into your model
prediction = loaded_model.predict(scaled_data)
print("Prediction:", prediction)
```
