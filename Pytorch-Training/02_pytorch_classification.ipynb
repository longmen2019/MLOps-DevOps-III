## Create a Model Class

Let's create a model class that:

- **Subclasses `nn.Module`**  
  Almost all PyTorch models inherit from `nn.Module`.

- **Creates two `nn.Linear` layers** in the constructor  
  These layers should handle the input and output shapes of **X** and **y**.

- **Defines a `forward()` method**  
  This method contains the forward-pass computation of the model.

- **Instantiates the model class**  
  Then sends the model to the target device (CPU or GPU).

```python
# 1. Construct a model class that subclasses nn.Module
class CircleModelV0(nn.Module):                           # define a custom neural network model
    def __init__(self):
        super().__init__()                                 # initialize the parent nn.Module class
        
        # 2. Create 2 nn.Linear layers capable of handling X and y input and output shapes
        self.layer_1 = nn.Linear(in_features=2,            # first layer: takes 2 inputs (x1, x2)
                                 out_features=5)           # outputs 5 features (hidden representation)
        
        self.layer_2 = nn.Linear(in_features=5,            # second layer: takes 5 hidden features
                                 out_features=1)           # outputs 1 value (binary prediction)
    
    # 3. Define a forward method containing the forward pass computation
    def forward(self, x):
        # Return the output of layer_2, a single feature, the same shape as y
        return self.layer_2(self.layer_1(x))               # pass x → layer_1 → layer_2

# 4. Create an instance of the model and send it to target device
model_0 = CircleModelV0().to(device)                       # move model to CPU or GPU
model_0
```

```python
# Calculate the accuracy (a classification metric)
def accuracy_fn (y_true, y_pred):
    correct = torch.eq(y_true, y_pred).sum().item() # torch.eq() calculates where two tensors are equal 
    acc = (correct / len(y_pred)) * 100
    return acc 
```

### 3.2 Building a training and testing loop

```python
torch.manual_seed(42)                                      # set random seed for reproducibility

# Set the number of epochs  
epochs = 100                                               # number of times the model will see the full dataset

# Put data to target device  
X_train, y_train = X_train.to(device), y_train.to(device)  # move training data to CPU/GPU
X_test, y_test = X_test.to(device), y_test.to(device)      # move test data to CPU/GPU

# Building training and evaluation loop 
for epoch in range(epochs):                                # loop through the dataset multiple times
    model_0.train()                                        # set model to training mode
    y_logits = model_0(X_train).squeeze()                  # forward pass: raw outputs (logits)
    y_pred = torch.round(torch.sigmoid(y_logits))          # convert logits → probs → 0/1 predictions

    loss = loss_fn(y_logits, y_train)                      # compute training loss
    acc = accuracy_fn(y_true=y_train, y_pred=y_pred)       # compute training accuracy
    
    optimizer.zero_grad()                                  # reset gradients before backprop

    loss.backward()                                        # compute gradients via backpropagation

    optimizer.step()                                       # update model parameters

    model_0.eval()                                         # set model to evaluation mode
    with torch.inference_mode():                           # disable gradient tracking for inference
        test_logits = model_0(X_test).squeeze()            # forward pass on test data
        test_pred = torch.round(torch.sigmoid(test_logits))# convert test logits → 0/1 predictions

        test_loss = loss_fn(test_logits, y_test)           # compute test loss

        test_acc = accuracy_fn(y_true=y_test,              # compute test accuracy
                               y_pred=test_pred)

    if epoch % 10 == 0:                                    # print progress every 10 epochs
        print(f"Epoch: {epoch} | Loss: {loss:.5f}, Accuracy: {acc:.2f}% | "
              f"Test Loss: {test_loss:.5f}, Test acc: {test_acc: .2f}%")
```

### 5.2 Adjusting model_1 to fit a straight line 

```python
# Same architecture as model_1 (but using nn.Sequential)
model_2 = nn.Sequential(                                   # build a model by stacking layers in order
    nn.Linear(in_features=1, out_features=10),             # first layer: map 1‑D input to 10 hidden units
    nn.Linear(in_features=10, out_features=10),            # second hidden layer: 10 → 10
    nn.Linear(in_features=10, out_features=1)              # output layer: produce a single prediction value
).to(device)                                                # move the entire model to CPU/GPU

model_2                                                     # display the model architecture
```

```python
# Loss and optimizer  
loss_fn = nn.L1Loss()                                   # use MAE (L1) loss to measure how far predictions are from targets
optimizer = torch.optim.SGD(model_2.parameters(),       # create optimizer to update model_2's parameters
                            lr=0.1)                     # set learning rate controlling step size during training
```

```python
# Train the model
torch.manual_seed(42)                                          # set random seed so training is reproducible

# Set the number of epochs
epochs = 1000                                                  # number of full passes through the training data

# Put data to target device
X_train_regression, y_train_regression = X_train_regression.to(device), y_train_regression.to(device)
                                                               # move training data to CPU/GPU
X_test_regression, y_test_regression = X_test_regression.to(device), y_test_regression.to(device)
                                                               # move test data to CPU/GPU

for epoch in range(epochs):                                    # loop through training for each epoch
    ### Training 
    # 1. Forward pass
    y_pred = model_2(X_train_regression)                       # compute model predictions on training data
    
    # 2. Calculate loss (no accuracy since it's a regression problem, not classification)
    loss = loss_fn(y_pred, y_train_regression)                 # compute training loss using L1/MAE

    # 3. Optimizer zero grad
    optimizer.zero_grad()                                      # clear previous gradients

    # 4. Loss backwards
    loss.backward()                                            # compute gradients via backpropagation

    # 5. Optimizer step
    optimizer.step()                                           # update model parameters

    ### Testing
    model_2.eval()                                             # switch model to evaluation mode
    with torch.inference_mode():                               # disable gradient tracking for inference
      # 1. Forward pass
      test_pred = model_2(X_test_regression)                   # compute predictions on test data

      # 2. Calculate the loss 
      test_loss = loss_fn(test_pred, y_test_regression)        # compute test loss

    # Print out what's happening
    if epoch % 100 == 0:                                       # print progress every 100 epochs
        print(f"Epoch: {epoch} | Train loss: {loss:.5f}, Test loss: {test_loss:.5f}")
```

### Building a model with non-linearity

```python
# Build model with non-linear activation function
from torch import nn                                           # import PyTorch's neural network module

class CircleModelV2(nn.Module):                                # define a neural network class for nonlinear classification
    def __init__(self):                                        # constructor to set up layers and activations
        super().__init__()                                     # initialize the parent nn.Module class
        self.layer_1 = nn.Linear(in_features=2, out_features=10)   # first layer: map 2‑D input to 10 hidden units
        self.layer_2 = nn.Linear(in_features=10, out_features=10)  # second hidden layer: 10 → 10
        self.layer_3 = nn.Linear(in_features=10, out_features=1)   # output layer: produce a single logit
        self.relu = nn.ReLU()                                  # define ReLU activation to introduce nonlinearity
        # Can also put sigmoid in the model 
        # This would mean you don't need to use it on the predictions
        # self.sigmoid = nn.Sigmoid()                           # optional sigmoid layer for output activation

    def forward(self, x):                                       # define how data flows through the model
       # Intersperse the ReLU activation function between layers
       return self.layer_3(self.relu(self.layer_2(self.relu(self.layer_1(x)))))
                                                                # apply layer → ReLU → layer → ReLU → output

model_3 = CircleModelV2().to(device)                           # create model instance and move it to CPU/GPU
print(model_3)                                                 # display the model architecture
```

```python
import torch                     # import PyTorch for tensor operations and model training
import torch.nn as nn            # import neural‑network module so nn.* layers and losses work

# Setup loss and optimizer 
loss_fn = nn.BCEWithLogitsLoss()              # binary classification loss that combines sigmoid + BCE in one stable function
optimizer = torch.optim.SGD(model_3.parameters(), lr=0.1)  
                                              # create SGD optimizer to update model_3's parameters with learning rate 0.1
```

### Training a model with non-linearity

```python
# Fit the model
torch.manual_seed(42)                                          # set random seed so training results are reproducible
epochs = 1000                                                  # number of full training iterations over the dataset

# Put all data on target device
X_train, y_train = X_train.to(device), y_train.to(device)      # move training data to CPU/GPU
X_test, y_test = X_test.to(device), y_test.to(device)          # move test data to CPU/GPU

for epoch in range(epochs):                                    # loop through training for each epoch
    # 1. Forward pass
    y_logits = model_3(X_train).squeeze()                      # compute raw logits from the model
    y_pred = torch.round(torch.sigmoid(y_logits))              # convert logits → probabilities → 0/1 predictions
    
    # 2. Calculate loss and accuracy
    loss = loss_fn(y_logits, y_train)                          # compute training loss using logits
    acc = accuracy_fn(y_true=y_train,                          # compute training accuracy
                      y_pred=y_pred)
    
    # 3. Optimizer zero grad
    optimizer.zero_grad()                                      # clear previous gradients

    # 4. Loss backward
    loss.backward()                                            # compute gradients via backpropagation

    # 5. Optimizer step
    optimizer.step()                                           # update model parameters

    ### Testing
    model_3.eval()                                             # switch model to evaluation mode
    with torch.inference_mode():                               # disable gradient tracking for inference
      # 1. Forward pass
      test_logits = model_3(X_test).squeeze()                  # compute logits on test data
      test_pred = torch.round(torch.sigmoid(test_logits))      # convert logits → probabilities → labels
      
      # 2. Calculate loss and accuracy
      test_loss = loss_fn(test_logits, y_test)                 # compute test loss
      test_acc = accuracy_fn(y_true=y_test,                    # compute test accuracy
                             y_pred=test_pred)

    # Print out what's happening
    if epoch % 100 == 0:                                       # print progress every 100 epochs
        print(f"Epoch: {epoch} | Loss: {loss:.5f}, Accuracy: {acc:.2f}% | "
              f"Test Loss: {test_loss:.5f}, Test Accuracy: {test_acc:.2f}%")
```

### 8. Putting things together by building a multi-class Pytorch model 

### 8.1 Creating Multi-Class Classification Data  

```python
# import dependencies 
import torch                                                # import PyTorch for tensor operations
import matplotlib.pyplot as plt                             # import matplotlib for plotting
from sklearn.datasets import make_blobs                     # import function to generate clustered data
from sklearn.model_selection import train_test_split        # import utility to split data into train/test sets

# Set the hyperparameters for data creation 
NUM_CLASSES = 4                                             # number of distinct classes (clusters)
NUM_FEATURES = 2                                            # number of features per sample (2D points)
RANDOM_SEED = 42                                            # seed for reproducibility

# 1. Create multi-class data  
X_blob, y_blob = make_blobs(n_samples=1000,                 # generate 1000 samples
                            n_features=NUM_FEATURES,        # each sample has 2 features
                            centers=NUM_CLASSES,            # create 4 cluster centers
                            cluster_std=1.5,                # spread of each cluster
                            random_state=RANDOM_SEED)       # ensure reproducible dataset

# 2. Turn data into tensors 
X_blob = torch.from_numpy(X_blob).type(torch.float)         # convert features to float32 tensor
y_blob = torch.from_numpy(y_blob).type(torch.LongTensor)    # convert labels to integer tensor
print(X_blob[:5], y_blob[:5])                               # preview first 5 samples and labels

# 3. Split into train and test sets 
X_blob_train, X_blob_test, y_blob_train, y_blob_test = train_test_split(
    X_blob,                                                 # input features
    y_blob,                                                 # labels
    test_size=0.2,                                          # allocate 20% for testing
    random_state=RANDOM_SEED)                               # reproducible split

# 4. Plot data 
plt.figure(figsize=(10,7))                                  # create a large figure for visualization
plt.scatter(X_blob[:,0], X_blob[:,1], c=y_blob,             # plot points colored by class label
            cmap=plt.cm.RdYlBu);                            # use red‑yellow‑blue colormap
```

### 8.2 Building a multi-class classification model in PyTorch 

```python
# Create devnice agnostic code 
device = "cuda" if torch.cuda.is_available() else "cpu"
device 
```

```python
from torch import nn                                            # import PyTorch's neural‑network module

# Build model
class BlobModel(nn.Module):                                     # define a model class for multi‑class classification
    def __init__(self, input_features, output_features, hidden_units=8):
        """Initializes all required hyperparameters for a multi-class classification model.

        Args:
            input_features (int): Number of input features to the model.
            out_features (int): Number of output features of the model
              (how many classes there are).
            hidden_units (int): Number of hidden units between layers, default 8.
        """
        super().__init__()                                      # initialize the parent nn.Module class
        self.linear_layer_stack = nn.Sequential(                # create a stack of linear layers
            nn.Linear(in_features=input_features,               # first layer: input → hidden
                      out_features=hidden_units),
            # nn.ReLU(),                                        # optional nonlinearity (try enabling to see effect)
            nn.Linear(in_features=hidden_units,                 # second layer: hidden → hidden
                      out_features=hidden_units),
            # nn.ReLU(),                                        # optional nonlinearity
            nn.Linear(in_features=hidden_units,                 # final layer: hidden → number of classes
                      out_features=output_features),
        )
    
    def forward(self, x):                                       # define forward pass
        return self.linear_layer_stack(x)                       # pass input through the stacked layers

# Create an instance of BlobModel and send it to the target device
model_4 = BlobModel(input_features=NUM_FEATURES,               # specify number of input features (2)
                    output_features=NUM_CLASSES,               # specify number of output classes (4)
                    hidden_units=8).to(device)                 # move model to CPU/GPU
model_4                                                        # display model architecture
```

### 8.5 Creating a training loop and testing loop for a multi-class Pytorch model 

```python
# Fit the model
torch.manual_seed(42)                                          # set random seed so training is reproducible

# Set number of epochs
epochs = 100                                                   # train for 100 full passes through the dataset

# Put data to target device
X_blob_train, y_blob_train = X_blob_train.to(device), y_blob_train.to(device)
                                                               # move training data to CPU/GPU
X_blob_test, y_blob_test = X_blob_test.to(device), y_blob_test.to(device)
                                                               # move test data to CPU/GPU

for epoch in range(epochs):                                    # loop through each training epoch
    ### Training
    model_4.train()                                            # set model to training mode

    # 1. Forward pass
    y_logits = model_4(X_blob_train)                           # compute raw class logits from the model
    y_pred = torch.softmax(y_logits, dim=1).argmax(dim=1)      # convert logits → probabilities → predicted class labels
    
    # 2. Calculate loss and accuracy
    loss = loss_fn(y_logits, y_blob_train)                     # compute training loss using logits
    acc = accuracy_fn(y_true=y_blob_train,                     # compute training accuracy
                      y_pred=y_pred)

    # 3. Optimizer zero grad
    optimizer.zero_grad()                                      # clear previous gradients

    # 4. Loss backwards
    loss.backward()                                            # compute gradients via backpropagation

    # 5. Optimizer step
    optimizer.step()                                           # update model parameters

    ### Testing
    model_4.eval()                                             # switch model to evaluation mode
    with torch.inference_mode():                               # disable gradient tracking for inference
      # 1. Forward pass
      test_logits = model_4(X_blob_test)                       # compute logits on test data
      test_pred = torch.softmax(test_logits, dim=1).argmax(dim=1)
                                                               # convert logits → probabilities → predicted labels
      
      # 2. Calculate test loss and accuracy
      test_loss = loss_fn(test_logits, y_blob_test)            # compute test loss
      test_acc = accuracy_fn(y_true=y_blob_test,               # compute test accuracy
                             y_pred=test_pred)

    # Print out what's happening
    if epoch % 10 == 0:                                        # print progress every 10 epochs
        print(f"Epoch: {epoch} | Loss: {loss:.5f}, Acc: {acc:.2f}% | "
              f"Test Loss: {test_loss:.5f}, Test Acc: {test_acc:.2f}%")
```
