### One of the most common errors in deep learning: shape errors

```python
# Shape for matrix multiplications (we'll define two matrices and multiply them)

# Define a 3x2 tensor (matrix) with explicit integer values
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])

# Define another 3x2 tensor (matrix) with explicit integer values
tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])

# torch.mm is an alias for torch.matmul when used with 2D matrices
# Perform matrix multiplication between tensor_A (3x2) and tensor_B (3x2)
# ⚠️ This will actually raise an error because the inner dimensions do not align:
# tensor_A is (3x2) and tensor_B is (3x2), so multiplication is not valid.
torch.matmul(tensor_A, tensor_B)
```

```python
# To fix our Tensor shape issues, we can manipulate the shape of one of them using a transpose.
# A transpose switches the axes (rows become columns, columns become rows).
# tensor_B.T changes tensor_B from shape (3, 2) to shape (2, 3).
tensor_B.T 
```

```python
tensor_B
```

```python
tensor_B.T.shape
```

```python
# The matrix multiplication operation works when tensor_B is transposed,
# because transposing switches its shape from (3, 2) to (2, 3).
# This makes the inner dimensions align: tensor_A (3x2) × tensor_B.T (2x3) → result (3x3).

print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")

# tensor_A keeps the same shape (3, 2), tensor_B.T changes to (2, 3)
print(f"New shapes: tensor_A = {tensor_A.shape}, tensor_B.T = {tensor_B.T.shape}")

# Show the multiplication operation with aligned inner dimensions
print(f"Multiplying: {tensor_A.shape} @ {tensor_B.T.shape}  <- inner dimensions must match")

print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)

# Display the shape of the resulting matrix
print(f"\nOutput shape: {output.shape}")

# Final multiplication result (same as stored in 'output')
torch.matmul(tensor_A, tensor_B.T)
```

## Reshaping, Stacking, Squeezing, and Unsqueezing Tensors

- **reshape** — Reshapes an input tensor to a defined shape.
- **view** — Returns a view of an input tensor with a specified shape while sharing the same underlying memory.
- **stack** — Combines multiple tensors either vertically (`vstack`) or horizontally (`hstack`).
- **squeeze** — Removes all dimensions of size `1` from a tensor.
- **unsqueeze** — Adds a dimension of size `1` at a specified position in a tensor.
- **permute** — Returns a view of the input tensor with its dimensions rearranged (swapped) in a specified order.


```python
# Let's create a tensor
import torch

x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add an extra dimension 
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape 
```

```python
# Change the view
z = x.view(1,9)
z, z.shape
```

```python
# Changing z changes x (because a view of a tensor shares the same memory as the original)
z[:,0] = 5 
z, x 
```

```python
# Stack tensor on top of each other 
x_stacked = torch.stack([x,x,x,x], dim=0)
x_stacked
```

```python
# torch.squeeze() - removes all single-dimension entries from the tensor

print(f"Previous tensor : {x_reshaped}")        # Display the original reshaped tensor
print(f"Previous tensor : {x_reshaped}")        # Same print (duplicate) showing the tensor before squeezing

# Remove extra dimensions from x_reshaped
x_squeezed = x_reshaped.squeeze()               # Apply squeeze() to drop all dimensions of size 1

print(f"\nNew tensor: {x_squeezed}")            # Display the squeezed tensor
print(f"\nNew shape: {x_squeezed.shape}")       # Show the new shape after squeezing
```

```python
x_reshaped.shape
```

```python
x_reshaped.squeeze()
```

```python
x_reshaped.squeeze().shape 
```

```python
# torch.unsqueeze() - adds a single dimension to a tensor at a specific dimension index

print(f"Previous target: {x_squeezed}")          # Show the tensor before adding a new dimension
print(f"Previous shape: {x_squeezed.shape}")     # Display the shape before unsqueezing

# Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)       # Insert a new dimension at index 0

print(f"\nNew tensor: {x_unsqueezed}")           # Show the tensor after unsqueezing
print(f"\nNew shape: {x_unsqueezed.shape}")      # Display the updated shape with the new dimension
```

```python
# torch.permute - rearranges the dimensions of a tensor in a specified order

x_original = torch.rand(size=(224, 224, 3))      # Create a tensor shaped [height, width, color_channels]

# Permute the original tensor to rearrange the axis (dimension) order
x_permuted = x_original.permute(2, 0, 1)         # Move dim 2 → 0, dim 0 → 1, dim 1 → 2 (HWC → CHW)

print(f"Previous shape: {x_original.shape}")                     # Show the original tensor shape
print(f"\nNew_shape: {x_permuted.shape}")          # Display the new shape after permuting
```

## Reproducibility (removing randomness for consistent results)


```python
torch.rand(3,3)
```

```python
import torch                                   # Import PyTorch for tensor creation and operations

# create two random tensors 
random_tensor_A = torch.rand(3, 4)             # Create a 3×4 tensor with random values between 0 and 1
random_tensor_B = torch.rand(3, 4)             # Create another independent 3×4 random tensor

print(random_tensor_A, random_tensor_B)        # Print both tensors to inspect their values

print(random_tensor_A == random_tensor_B)      # Compare element‑wise to check where values are equal
```

```python
# Let's make some random but reproducible tensors 
import torch                                   # Import PyTorch for tensor operations

# Set the random seed  
RANDOM_SEED = 42                                # Define a fixed seed value for reproducibility
torch.manual_seed(RANDOM_SEED)                  # Apply the seed so random numbers become repeatable

random_tensor_C = torch.rand(3, 4)              # Create a 3×4 tensor with reproducible random values
random_tensor_D = torch.rand(3, 4)              # Create another 3×4 tensor with the next reproducible values

print(random_tensor_C)                          # Print the first random tensor
print(random_tensor_D)                          # Print the second random tensor
print(random_tensor_C == random_tensor_D)       # Compare tensors element‑wise for equality
```

```python
# Let's make some random but reproducible tensors 
import torch                                   # Import PyTorch for tensor operations

# Set the random seed  
RANDOM_SEED = 42                                # Define a fixed seed value for reproducibility
torch.manual_seed(RANDOM_SEED)                  # Apply the seed so random numbers become repeatable

random_tensor_C = torch.rand(3, 4)              # Generate a 3×4 tensor with reproducible random values

torch.manual_seed(RANDOM_SEED)                  # Reset the seed so the next tensor is identical to the first
random_tensor_D = torch.rand(3, 4)              # Generate another 3×4 tensor with the same reproducible values

print(random_tensor_C)                          # Print the first random tensor
print(random_tensor_D)                          # Print the second random tensor (identical to the first)
print(random_tensor_C == random_tensor_D)       # Compare tensors element‑wise for equality (all True)
```
